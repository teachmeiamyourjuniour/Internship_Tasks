```python
import copy
a=[[1,2],[3,4]]
b= copy.copy(a)
b[0][1]=99
print(a)
```

```python
#  copy.copy makes a shallow copy of the outer list, but the nested lists remain shared. Changing b[0][1] therefore also changes the nested list visible through a.
```

```python
import copy
a=[[1,2],[3,4]]
b= copy.deepcopy(a)
b[0][1]=99
print(a)
```

```python
# copy.deepcopy recursively copies the nested lists as well as the outer list. Changing a nested element in b does not affect a, so a keeps its original values.
```

```python
import copy
a=[[10],[20]]
b= a.copy()
a[1].append(30)
print(b)
```

```python
#  a.copy() makes a shallow copy: the outer lists differ, but their inner lists are shared. Appending 30 to a[1] is visible through b[1].
```

```python
import copy
a=[[10],[20]]
b=copy.deepcopy(a)
a[1].append(30)
print(b)
```

```python
# deepcopy creates independent inner lists. Appending 30 to a[1] does not change the nested list held by b.
```

```python
import copy
a=[[1,2],[3,4]]
b=copy.copy(a)
print(a is b)
print(a[0] is b[0])
```

```python
#  copy.copy creates a different outer list, so a is b is False. The nested list at index 0 is shared by both shallow copies, so a[0] is b[0] is True.
```

```python
def test(*args):
    print(args)
test(1,2)
```

```python
# *args collects positional arguments passed to a function into a tuple. Passing 1 and 2 makes args equal to (1, 2).
```

```python
def test(*args):
    print(type(args))
test("a","b","c")
```

```python
#   The *args parameter collects all positional arguments into a tuple, so type(args) reports tuple.
```

```python
def test(**kwargs):
    print(kwargs)
test(name="Ram",age=25)
```

```python
# **kwargs collects keyword arguments into a dictionary. The keyword names become keys and their supplied values become the corresponding values.
```

```python
def test(**kwargs):
    print(type(kwargs))
test(a=10,b=20)
```

```python
# **kwargs gathers keyword arguments into a dictionary, so type(kwargs) reports dict.
```
