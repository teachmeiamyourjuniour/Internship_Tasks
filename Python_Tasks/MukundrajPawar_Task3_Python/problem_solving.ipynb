```python
def word_frequency(sentence):
    words = sentence.split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    return freq

word_frequency( "python is easy and python is powerful")
```

```python
word_frequency( "hello world hello")
```

```python
def char_frequency(word):
    freq = {}
    for char in word:
        freq[char] = freq.get(char, 0) + 1
    return freq

char_frequency("banana")
```

```python
char_frequency("apple")
```

```python
def first_non_repeating(word):
    freq = {}
    for char in word:
        freq[char] = freq.get(char, 0) + 1

    for char in word:
        if freq[char] == 1:
            return char
    return None

first_non_repeating("swiss")
```

```python
first_non_repeating("aabbcc")
```

```python
def are_anagrams(s1, s2):
    return sorted(s1) == sorted(s2)
    
are_anagrams("listen", "silent")
```

```python
are_anagrams("triangle", "integral")
```

```python
def first_repeating(s):
    seen = set()
    for char in s:
        if char in seen:
            return char
        seen.add(char)
    return None

first_repeating("programming")
```

```python
first_repeating("abca")
```

```python
def move_zeros(lst):
    non_zeros = [x for x in lst if x != 0]
    zeros = [0] * (len(lst) - len(non_zeros))
    return non_zeros + zeros
move_zeros([0, 3, 0, 1, 2, 0, 4])
```

```python
move_zeros([1, 0, 2, 0, 3])
```

```python
def second_highest(lst):
    unique_nums = list(set(lst))
    if len(unique_nums) < 2:
        return None
    unique_nums.sort(reverse=True)
    return unique_nums[1]
second_highest([10, 20, 40, 30, 40, 20])
```

```python
second_highest([5, 5, 4, 3])
```

```python
def remove_duplicates(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result
remove_duplicates([4, 2, 4, 5, 2, 7, 5])
```

```python
remove_duplicates([1, 1, 2, 2, 3])
```

```python
def find_missing(nums, N):
    expected_sum = N * (N + 1) // 2
    actual_sum = sum(nums)
    return expected_sum - actual_sum
find_missing([1, 2, 4, 5, 6], 6)
```

```python
find_missing([1, 2, 3, 5], 5)
```

```python
def two_sum(nums, target):
    seen = set()
    for num in nums:
        needed = target - num
        if needed in seen:
            return [needed, num]
        seen.add(num)
    return []
two_sum([2, 7, 11, 15], 9)
```

```python
two_sum([3, 2, 4], 6)
```

```python
def closest_to_zero(lst):
    closest = lst[0]
    for num in lst:
        if abs(num) < abs(closest):
            closest = num
        elif abs(num) == abs(closest) and num > closest:
            closest = num
    return closest
closest_to_zero([-8, -3, 2, 5, -2])
```

```python
closest_to_zero([-5, 5, -1, 2])
```

```python
def common_elements(list1, list2):
    set2 = set(list2)
    result = []
    seen = set()
    for item in list1:
        if item in set2 and item not in seen:
            result.append(item)
            seen.add(item)
    return result
common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7])
```

```python
common_elements([1, 2, 2, 3], [2, 3, 4])
```

```python
def rotate_list(nums, K):
    if not nums:
        return nums
    K = K % len(nums)
    if K == 0:
        return nums
    return nums[-K:] + nums[:-K]
rotate_list([1, 2, 3, 4, 5], 2)
```

```python
rotate_list([1, 2, 3], 1)
```

```python
def max_consecutive_sum(lst):
    max_sum = float('-inf')
    for i in range(len(lst) - 1):
        current_sum = lst[i] + lst[i + 1]
        if current_sum > max_sum:
            max_sum = current_sum
    return max_sum

max_consecutive_sum([4, 7, 2, 9, 5])
```

```python
max_consecutive_sum([1, 5, 2, 8])
```

```python
def all_pairs_sum_target(nums, target):
    seen = set()
    pairs = set()
    for num in nums:
        needed = target - num
        if needed in seen:
            pair = tuple(sorted((needed, num)))
            pairs.add(pair)
        seen.add(num)
    return [list(p) for p in sorted(list(pairs))]

all_pairs_sum_target([2, 4, 3, 5, 7, 8, 1], 9)
```

```python
all_pairs_sum_target([1, 5, 3, 3, 4], 6)
```
